# Classroom Exam Score Prediction

Load the trained model included with this project and use sample study habits to predict an exam score. Run this notebook from the project folder so `best_model.pkl` can be found.

```python
from pathlib import Path

import joblib
import pandas as pd

model_path = Path("best_model.pkl")
if not model_path.exists():
    raise FileNotFoundError(f"Could not find the trained model: {model_path.resolve()}")

model = joblib.load(model_path)
print(f"Loaded model from {model_path.resolve()}")
```

## Make a prediction

Edit the sample values below. The input column names must match the features used to train the model.

```python
student_habits = pd.DataFrame([
    {
        "Attendance (%)": 85.0,
        "Daily Study Hours": 2.5,
        "Daily Screen Time (Hours)": 4.0,
        "Daily Sleep (Hours)": 7.0,
        "Daily Exercise (Hours)": 1.0,
    }
])

prediction = float(model.predict(student_habits)[0])
prediction = max(0.0, min(100.0, prediction))
print(f"Predicted exam score: {prediction:.1f}%")
```
